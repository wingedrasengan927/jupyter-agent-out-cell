## How it works

- The function `styleAgentOutCell` checks if a cell has the required metadata (`"agent_out": true`) and adds the css class for styling
- We add an event listener to the notebook panel that listens for a cell change event and triggers `styleAgentOutCell` if the change type is `add`
- When the notebook is initialised, we run `styleAgentOutCell` on each cell
- Essentially, when change type `add` is detected, we grab the newly added cell and apply `styleAgentOutCell`
